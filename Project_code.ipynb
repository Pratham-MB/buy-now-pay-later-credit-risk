```python
bnpl_data.drop(columns=['user_id'], axis=1, inplace=True)
#axis=1 means we are dropping a column, inplace=True means we are making the change in the original dataframe
#inplace=True means that the change will be made in the original dataframe, and we don't need to assign it to a new variable.
bnpl_data.head()
```

```python
bnpl_data['gender'] = bnpl_data['gender'].map({'Male': 1, 'Female': 0})
bnpl_data.head()
```

```python
import matplotlib.pyplot as plt
# Plotting histogram using matplotlib
plt.figure(figsize=(8, 5))
bnpl_data['risk_score'].hist(bins=20, edgecolor='black', color='skyblue')
plt.xlabel('risk_score')
plt.ylabel('Count')
plt.title('Histogram of risk_score')
plt.grid(False)
plt.show()
```

```python
import matplotlib.pyplot as plt

counts = bnpl_data['default_flag'].value_counts().sort_index()
plt.figure(figsize=(6, 4))
counts.plot(kind='bar', color=['skyblue', 'salmon'], edgecolor='black')
plt.xlabel('default_flag')
plt.ylabel('Count')
plt.title('Distribution of default_flag')
plt.xticks(rotation=0)
plt.grid(False)
```

```python
import seaborn as sns
plt.figure(figsize=(14, 10))

numeric_cols = bnpl_data.select_dtypes(include=['int64', 'float64']).columns
correlation_matrix = bnpl_data[numeric_cols].corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt
```

```python
fig, axes = plt.subplots(1, 2, figsize=(20, 6))

sns.countplot(data=bnpl_data, x='employment_type', hue='default_flag', ax=axes[0])
axes[0].set_title('Default Flag by Employment Type')
axes[0].tick_params(axis='x')
```

```python
import matplotlib.pyplot as plt

locations = bnpl_data['location'].unique()
fig, axes = plt.subplots(1, len(locations), figsize=(20, 5))

for i, loc in enumerate(locations):
    subset = bnpl_data[bnpl_data['location'] == loc]
    grouped = subset.groupby(['employment_type', 'default_flag']).size().unstack(fill_value=0)
    grouped.plot(kind='bar', stacked=True, ax=axes[i])
    axes[i].set_title(f'Employment Type and Default Flag in {loc}')
    axes[i].set_xlabel('Employment Type')
    axes[i].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
